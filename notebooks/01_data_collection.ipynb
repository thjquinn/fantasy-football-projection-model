```python
import nflreadpy as nfl

seasons = list(range(2016,2026))

player_stats = nfl.load_player_stats(
    seasons=seasons,
    summary_level="reg"
)

player_stats.head()
```

```python
print(player_stats.shape)
print(player_stats.columns)
```

```python
core_columns = [
    "player_id",
    "player_display_name",
    "position",
    "season",
    "recent_team",
    "games",

    # Passing
    "attempts",
    "passing_yards",
    "passing_tds",
    "passing_interceptions",
    "passing_2pt_conversions",

    # Rushing
    "carries",
    "rushing_yards",
    "rushing_tds",
    "rushing_2pt_conversions",

    # Receiving
    "targets",
    "receptions",
    "receiving_yards",
    "receiving_tds",
    "receiving_2pt_conversions",

    # Fumbles
    "fumbles_lost_total",
    "sack_fumbles_lost",
    "rushing_fumbles_lost",
    "receiving_fumbles_lost",

    # Special Teams
    "special_teams_tds",

    # Advanced Metrics
    "target_share",
    "air_yards_share",
    "wopr",

    # Existing Fantasy Points (For Comparison)
    "fantasy_points",
    "fantasy_points_ppr"
]

player_stats_core = player_stats.select(core_columns)

player_stats_core.head()
```

```python
print(player_stats_core.shape)
print(player_stats_core.columns)
```

```python
#Filter to only include QB, RB, WR, TE

fantasy_positions = ["QB","RB","WR","TE"]

player_stats_core = player_stats_core.filter(
    player_stats_core["position"].is_in(fantasy_positions)
)

player_stats_core.head()
```

```python
#Checking to see how many players by position

player_stats_core.group_by("position").len().sort("position")
```

```python
print(player_stats_core.shape)

player_stats_core.group_by("season").len().sort("season")
```

```python
#Checking for duplicate players 

duplicates = (
    player_stats_core
    .group_by(["player_id","season"])
    .len()
    .filter(player_stats_core.group_by(["player_id","season"]).len()["len"] > 1)
)

duplicates
```

```python
print("Duplicate player-seasons:", duplicates.height)
```

```python
# Fantasy Score Self Calculation

import polars as pl

player_stats_core = player_stats_core.with_columns(
    (
        pl.col("passing_yards") / 25
        + pl.col("passing_tds") * 4
        - pl.col("passing_interceptions") * 2

        + pl.col("rushing_yards") / 10
        + pl.col("rushing_tds") * 6

        + pl.col("receiving_yards") / 10
        + pl.col("receptions")
        + pl.col("receiving_tds") * 6

        + pl.col("passing_2pt_conversions") * 2
        + pl.col("rushing_2pt_conversions") * 2
        + pl.col("receiving_2pt_conversions") * 2

        + pl.col("special_teams_tds") * 6

        - (
            pl.col("sack_fumbles_lost")
            + pl.col("rushing_fumbles_lost")
            + pl.col("receiving_fumbles_lost")
        ) * 2

    ).alias("calculated_fantasy_points_ppr")
)
```

```python
#Comparing self calculated score with actual fantasy points

comparison = player_stats_core.select([
    "player_display_name",
    "position",
    "fantasy_points_ppr",
    "calculated_fantasy_points_ppr"
]).with_columns(
    (
        pl.col("calculated_fantasy_points_ppr")
        - pl.col("fantasy_points_ppr")
    ).alias("difference")
)

comparison.sort(
    pl.col("difference").abs(),
    descending=True
).head(20)
```

```python
comparison.select(
    pl.col("difference").abs().mean().alias("mean_absolute_difference"),
    pl.col("difference").abs().max().alias("max_absolute_difference")
)
```

```python
comparison.filter(
    pl.col("difference") != 0
).sort(
    pl.col("difference").abs(),
    descending=True
)
```

```python
problem_players = (
    comparison
    .filter(pl.col("difference") != 0)
    .select("player_display_name")
)

player_stats_core.join(
    problem_players,
    on="player_display_name",
    how="inner"
).select([
    "player_display_name",
    "position",
    "passing_2pt_conversions",
    "rushing_2pt_conversions",
    "receiving_2pt_conversions",
    "special_teams_tds",
    "fumbles_lost_total",
    "fantasy_points_ppr",
    "calculated_fantasy_points_ppr"
])
```

```python
player_stats_core.write_csv("../data/raw/player_stats_2016_2025.csv")
```
